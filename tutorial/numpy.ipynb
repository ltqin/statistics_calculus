numpy通常以np的名字导入

```python
import numpy as np
```

numpy用数组array来进行计算  
现在我们准备一个numpy的array，它的元素是[1,2,3,4,5]

```python
arr = np.array([1, 2, 3, 4, 5])
```

可以通过shape查看它的大小

```python
arr.shape
```

对array做加法运算

```python
arr + 2
```

所有的元素都被+2了 
我们称之为广播

不仅仅是加法运算，其他运算也会发生同样的事情  
我们试着操作一下

```python
np.sqrt(arr)
```

和列表一样，你可以在切片中提取元素  

```python
arr[:2]
```

```python
arr[3:5]
```

我们经常使用numpy的arange函数  
可以使用arange函数来创建带有[0,1,2,3,4]元素的array

```python
np.arange(5)
```

可以在arange函数中指定start, end, step  
[2, 4, 6, 8]这个以数字2递增的array可以是这样的

```python
np.arange(2, 10, 2)
```

我们也经常使用linspace函数  
linspace函数也是制作从开始到结束的array的函数，但不是step，而是指定元素的数量  
现在，我们将0到3分成11个array

```python
np.linspace(0, 3, 11)
```

二维的array也很容易制作

```python
arr2 = np.array([[1, 3],
                 [5, 7]])
arr2
```

```python
arr2.shape
```

可以用save函数保存array  
第一个参数是保存的名字，第二个参数是保存的array

```python
np.save('test.npy', arr2)
```

它可以通过load函数来读取

```python
np.load('test.npy')
```
