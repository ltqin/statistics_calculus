### 字符串的基础

字符串用单引号框住

```python
'Hello World!'
```

本书采用的是单引号，同样适用于双引号

```python
"Hello World!"
```

你也可以用同样的方式来处理日语

```python
'你好世界！'
```

也可以用print语句输出

```python
print('Hello', 'World')
```

```python
a = 'Hello'
b = 'World'
```

字符串可以用+组合

```python
a + b
```

```python
a[:2]
```

### f-strings

在Python3.6中，你可以使用f-strings

```python
a = 'test'
b = 3
```

f-strings可以直接在字符串中输出变量

```python
f'{a} is {b}'
```

```python
c = 100 / 7
c
```

可以通过以下方式来调整位数

```python
f'{c:.2f}'
```

# 列表

## 列表的基础

Python的列表用[]框住

```python
[2, 3, 4]
```

列表中的元素可以不是同一类型

```python
[2, '字符串', [2, 3]]
```

通过将列表嵌套，可以创建多维数组

```python
[[1, 2], [2, 3]]
```

## 列表的运算

```python
a = [1, 2, 3]
b = [3, 4, 5, 6]
```

列表的合并

```python
a + b
```

列表的追加

```python
a.append(3)
a
```

切片

```python
b[:3]
```

```python
b[1:3]
```

# 列表内涵标记

比如，你想列出一个从0到8的偶数列表

```python
a = []
for i in range(5):
    a.append(2 * i)
    
a
```

```python
[2 * i for i in range(5)]
```

```python
[i for i in range(10) if i % 2 == 0]
```

```python
[i if i%2 == 0 else 1 for i in range(10)]
```
